## **Theory: Gradient Descent**

**Cost Function (MSE):**  
$$ J = \frac{1}{2m} \sum (\hat{y} - y)^2 $$

**Update Rules:**  
$$ b := b - \alpha \cdot \frac{\partial J}{\partial b} $$
$$ m := m - \alpha \cdot \frac{\partial J}{\partial m} $$

## **Gradient Descent Class**

```python
class GDRegressor:

    def __init__(self,learning_rate=0.01,epochs=1000):
      self.learning_rate=learning_rate
      self.epochs=epochs
      self.m=0

      self.b=0


      def fit(self, X, y):
        X=X.flatten()
        n=len(X)


        for _ in range(self.epochs):
          y_pred=self.m*X+self.b# Step 1: Predict
            
           # Step 2: Calculate Gradients
          dm=(-2/n)*np.sum((y-y_pred)*X)
          db=(-2/n)*np.sum(y-y_pred)
           # Step 3: Update Parameters
          self.n=self.n-self.learning_rate*dm
          self.b=self.b-self.leaning_rate*db

      print("Training Completed")
      print(f"Slope (m)     : {self.m:.4f}")
      print(f"Intercept (b) : {self.b:.4f}")

      def predict(self, X):
         X=X.flatten()
         return self.n*X+self.b



```

# Adjusted R² Score

R² Score measures how much of the variance in the target variable is explained by the model.

However, R² has one limitation:

- Adding more features to a model will **never decrease** the R² score, even if those features are irrelevant.

Adjusted R² solves this problem by penalizing the model for adding unnecessary features.

### Formula

Adjusted R² = 1 - ((1 - R²) × (n - 1)) / (n - p - 1)

Where:
- R² = Coefficient of Determination
- n = Number of observations (samples)
- p = Number of independent variables (features)

A higher Adjusted R² indicates that the added features genuinely improve the model.
