```python
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv("train.csv")
```

# Categorical DATA
a. CountPlot

```python
sns.countplot(x='Pclass', data=df)
```

# b.Pie Chart


```python
df['Pclass'].value_counts().plot(kind='pie',autopct='%.2f')
```

# NUmerical Data

### 1. Histogram

```python
import matplotlib.pyplot as plt
plt.hist(df['Age'],bins=8)
```

### b. Distplot- Probability destination function- it tells probability

```python
sns.distplot(df['Age'])
```

# Box plot

### detect potential outliers  
if u have noise and outliers in data

```python
sns.boxplot(x=df['Age'],data=df)
```
