# What is Exploratory Data Analysis (EDA)?

Exploratory Data Analysis (EDA) is the process of examining, cleaning, summarizing, and visualizing data before building Machine Learning models.

EDA helps us understand:
- Structure of the dataset
- Data types
- Missing values
- Duplicate records
- Outliers
- Feature distributions
- Relationships among variables

EDA is often called the first and most important step in Data Analysis and Machine Learning.

### Why is EDA Important?

Imagine receiving marks of 5,000 students without any explanation.

Questions that immediately arise are:

- How many students are there?
- Are there missing values?
- What is the average score?
- Are there any incorrect entries?
- Are there duplicate records?
- Is the data balanced?

EDA answers these questions before model building.

### EDA Workflow

A typical EDA process includes:

1. Load the dataset
2. Inspect the data
3. Understand data types
4. Check missing values
5. Remove duplicates
6. Generate descriptive statistics
7. Visualize distributions
8. Detect outliers
9. Study feature relationships

### 6.4 Measures of Central Tendency

Central tendency refers to the center of a dataset.

The three common measures are:
1. Mean
2. Median
3. Mode

#### Mean

The Mean is the arithmetic average of all observations.

Formula: $$\text{Mean} = \frac{\Sigma x}{n}$$

#### Median

Median is the middle value after sorting. Unlike Mean, it is less affected by outliers.

Formula:
- Case 1: Odd Number of Observations
$$
\text{Median} = x_{\frac{n+1}{2}}
$$
- Case 2: Even Number of Observations
$$
\text{Median} =
\frac{x_{\frac{n}{2}} + x_{\frac{n}{2}+1}}{2}
$$


#### Mode

Mode is the most frequently occurring value. Unlike Mean and Median, Mode can also be used for categorical variables.

### 6.5 Measures of Dispersion

**What is Dispersion?** Measures of Dispersion describe how spread out the data values are from the center.

Two datasets may have the same mean but completely different spreads.

Example:
- Dataset A = [48, 49, 50, 51, 52]
- Dataset B = [10, 30, 50, 70, 90]

Both datasets have: `Mean = 50`. But Dataset B has much greater variation. Dispersion helps quantify this variation.

---

#### Types of Dispersion

We commonly use:
1. Range
2. Variance
3. Standard Deviation
4. Interquartile Range (IQR)

#### 1. Range

Range is the difference between the maximum and minimum values.

Formula: $$\text{Range} = Maximum − Minimum$$

```python
# Range of Passenger Age
age_range=df['Age'].max()-df['Age'].min()
age_range
```

```python
# Range of Fare
fare_range=df['Fare'].max()-df['Fare'].min()
fare_range
```

#### 4. Interquartile Range (IQR)

A *percentile* is a statistical measure which tells us how a value compares with the rest of the data. For example, if a student's score is in the 80th percentile, it means the student scored higher than or equal to 80% of the students, while only 20% scored higher than her/him.

| Percentile       | Meaning                                 |
| ---------------- | --------------------------------------- |
| 0th              | Minimum value                           |
| 25th (Q1)        | 25% of values lie below this point      |
| 50th (Median/Q2) | Half of the values lie below this point |
| 75th (Q3)        | 75% of values lie below this point      |
| 100th            | Maximum value                           |

IQR measures the spread of the middle 50% of observations. It is less affected by outliers.

Formula: `IQR = Q3 − Q1`

Where,
- Q1 = 25th Percentile
- Q3 = 75th Percentile

# Introduction to Missing Values

**What are Missing Values?** Missing values occur when information is unavailable.

Examples:
- Passenger Age not recorded.
- Embarked location missing.
- Deck information unavailable.

Missing values appear as: `NaN`

#### Why Handle Missing Values?

Machine Learning algorithms cannot work properly with missing data.

Missing values may:
- Reduce model accuracy.
- Cause training errors.
- Introduce bias.

### Method 1: Constant Value Imputation

Replace missing values with a fixed value.

```python
# Replace Age with 0
df_const=df.copy()
df_const['Age']=df_const['Age'].fillna(0)
df_const['Age'].isnull().sum()
```

```python
# Replace Embarked with "Unknown"
df_const['Embarked']=df_const["Embarked"].fillna("Unknown")
```

### When to Use?

Useful when:
- Missing itself carries meaning.
- A default category is acceptable.

### Method 2: Mean Imputation

Suitable for normally distributed numerical data.

```python
# Mean of Age
mean_age=df['Age'].mean()
mean_age
```

```python
# Fill Missing Age
df['Age'].fillna(df['Age'].mean())

```

#### Advantages
- Simple
- Preserves sample size

#### Limitations
- Affected by outliers.

#### Try Yourself

### Method 3: Median Imputation

Median works better for skewed distributions.

```python
# Fill Missing Age
df['Age'].fillna(df['Age'].median())
```

#### Why Median?

Age and Fare often contain outliers. Median is less affected by extreme values.

### Comparison of Imputation Methods

| Method   | Best For                            | Affected by Outliers? |

| Median   | Skewed numerical data               | No                    |
