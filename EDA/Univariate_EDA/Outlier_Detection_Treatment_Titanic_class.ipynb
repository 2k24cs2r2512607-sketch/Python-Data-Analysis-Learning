## Method : Detecting Outliers Using IQR (Interquartile Range)

**What is IQR (Interquartile Range)?** The Interquartile Range (IQR) is a statistical measure that represents the spread of the middle 50% of the data. It measures the distance between the first quartile (Q1) and the third quartile (Q3).

In simple words, IQR tells us how spread out the central half of the dataset is, while ignoring extreme values (outliers). Because it focuses only on the middle 50% of the data, IQR is less affected by outliers than measures like the range or standard deviation.

Formula: `IQR = Q3 − Q1`

Where,
- Q1 = 25th Percentile
- Q3 = 75th Percentile

An observation is considered an *outlier* if `Value < Q1 − 1.5 × IQR` or `Value > Q3 + 1.5 × IQR`.

## Method : Detecting Outliers Using Z-Score

**What is Z-Score?** A Z-score (also called the *Standard Score*) is a statistical measure that indicates how many standard deviations a data point is away from the mean (average) of a dataset.

In simple words, It tells us how far a value is from the average (or mean) value of the dataset.

$$z = \frac{x - \mu}{\sigma}$$

Where,
- x = Data value
- μ (mu) = Mean of the dataset
- σ (sigma) = Standard deviation

Observation:
- A positive Z-score means the value is above the mean.
- A negative Z-score means the value is below the mean.
- A Z-score of 0 means the value is exactly equal to the mean.

Generally,
- $Z > 3$ → Outlier
- $Z < -3$ → Outlier

```python
df['fare_score']=(df['Fare']-df['Fare'].mean())/df['Fare'].std()
df
```
