##Q1. What is Feature Scaling? Why is it necessary?
Answer

Feature scaling is the process of transforming numerical features so that they are on a similar scale without changing their underlying distribution or relationships.

Many machine learning algorithms calculate distances, gradients, or coefficients during training. If one feature has values much larger than another, it can dominate the learning process and negatively affect model performance.

For example, consider two features from the Titanic dataset:

Feature	Approximate Range
Age	0.42–80 years
Fare	0–512 pounds

Since Fare has a much larger range than Age, algorithms such as K-Nearest Neighbors (KNN), Support Vector Machines (SVM), Logistic Regression, Linear Regression (with gradient descent), Principal Component Analysis (PCA), and Neural Networks may assign disproportionate importance to Fare.

Scaling ensures that each feature contributes more equally during model training.

Titanic columns that benefit from scaling
Age
Fare

Both are continuous numerical variables with different ranges and units.

# 1. Feature Scaling
**What is Feature Scaling?** Feature Scaling is the process of transforming numerical features into a common scale without changing their relationships.

Example

| Feature | Value |
|---------|--------|
| Age | 25 |
| Salary | 6,50,000 |

Notice that:
- Age ranges between 18–60
- Salary ranges between ₹20,000–₹20,00,000

Since Salary has much larger values, many Machine Learning algorithms give it more importance.

Feature Scaling solves this problem.

### Why is Feature Scaling Important?

Feature scaling is especially important for algorithms that calculate distances or gradients.

Examples include:
- K-Nearest Neighbors (KNN)
- K-Means Clustering
- Support Vector Machine (SVM)
- Logistic Regression
- Linear Regression (Gradient Descent)
- Neural Networks
- Principal Component Analysis (PCA)

Note: *Algorithms like **Decision Trees** and **Random Forest** usually do not require feature scaling.*

#### Example Without Scaling

Suppose we have:

| Height (cm) | Income (₹) |
|-------------|------------|
| 170 | 50000 |
| 175 | 55000 |
| 180 | 800000 |

The income values dominate the calculations because they are much larger.

#### Types of Feature Scaling

There are several commonly used methods:
- Standard Scaling or Standardization (Z-Score Scaling)
- Min-Max Scaling or Normalization
- Robust Scaling
- Log Transformation

#### Normalization vs Standardization
| Normalization                                                 | Standardization                                                  |
| ------------------------------------------------------------- | ---------------------------------------------------------------- |
| Scales values between **0 and 1**                             | Scales values around **mean = 0** and **standard deviation = 1** |
| Formula: $$\displaystyle x'=\frac{x-x_{min}}{x_{max}-x_{min}}$$ | Formula: $$\displaystyle z=\frac{x-\mu}{\sigma}$$               |
| Sensitive to outliers                                         | Less sensitive to outliers                                       |
| Used for KNN, K-Means, Neural Networks                        | Used for Regression, SVM, PCA                                    |


Let's study each of them.

## 1.1 Standard Scaler or Standardization
Standardization transforms the data using:

Formula: $$Z = \frac{X - \mu}{\sigma}$$
where

x = original value

μ = mean

σ = standard deviation

After transformation:

Mean ≈ 0

Standard deviation ≈ 1

Advantages
Centers the data around zero.

Preserves outliers.

Suitable for algorithms assuming normally distributed inputs.

## 1.3. Robust Scaler

Robust Scaling uses:
- Median
- Interquartile Range (IQR)

instead of *Mean* and *Standard Deviation*.

Formula:
$$X_{scaled} = \frac{X - Median}{IQR}$$

#### Why Robust Scaler?

Standard Scaler is affected by outliers. Robust Scaler ignores extreme values and scales using the middle 50% of the data.

For example, Data = [10, 12, 15, 18, 20, 500]. Since 500 is an outlier, Standard Scaling becomes distorted. Robust Scaling performs much better.

```python
# Create a copy of the dataset
df_robust = df.copy()

robust_scaler = RobustScaler()
df_robust[["age", "fare"]] = robust_scaler.fit_transform(df[["age", "fare"]])

df_robust.head()
```

## 1.4. Log Transformation

Log Transformation compresses large values and reduces skewness. Instead of using $x$, we use: $\log(x)$ or $\log(x+1)$.

The $+1$ avoids issues when the value is $0$.

Machine learning models perform mathematical operations such as computing distances, dot products, and gradients. Raw text values like "male", "female", or "S" cannot be used directly in these calculations.

Encoding allows models to interpret categorical information numerically while preserving its meaning.

##Q10. Label Encoding vs One-Hot Encoding on Pclass
Label Encoding

1 → 1

2 → 2

3 → 3

One-Hot Encoding

Pclass_1

Pclass_2

Pclass_3

Which is better?

Since Pclass has a natural order:

First Class
Second Class
Third Class

an ordinal representation is meaningful. Therefore, Label Encoding (or simply keeping the existing integer values) is generally more appropriate because it preserves the inherent ranking of the classes.

If there were no natural ordering (e.g., colors or cities), One-Hot Encoding would be the preferred choice.

### Label Encoding vs One-Hot Encoding
| Feature  | Label Encoding       | One-Hot Encoding   |
| -------- | -------------------- | ------------------ |
| Output   | Single Column        | Multiple Columns   |
| Ordering | Introduces order     | No order           |
| Memory   | Low                  | Higher             |
| Best For | Binary/Target Labels | Nominal Categories |



### Choosing the Right Technique
| Situation                           | Recommended Method |
| ----------------------------------- | ------------------ |
| Normally distributed numerical data | Standard Scaler    |
| Values between 0–1 needed           | Min-Max Scaler     |
| Many outliers                       | Robust Scaler      |
| Highly skewed numerical data        | Log Transformation |
| Binary categorical variable         | Label Encoding     |
| Nominal categorical variable        | One-Hot Encoding   |
